```python
import pandas as pd
import numpy as np

file_path = '/content/drive/MyDrive/Data Tugas 1 Penambangan Data/Data Penjualan E-Commerce.csv'

data = pd.read_csv(file_path)

data.head()
```

```python
print("Jumlah baris dan kolom:", data.shape)

print("\nNama-nama variabel:")
print(data.columns.tolist())

print("\nInformasi dataset:")
data.info()
```

```python
# Menghitung jumlah missing value setiap kolom

data.isnull().sum()
```

```python
# Memeriksa missing value yang masih berbentuk teks

missing_tokens = ['-', 'unknown', 'n/a', '']

for kolom in data.columns:
    jumlah = data[kolom].astype('string').str.strip().str.lower().isin(missing_tokens).sum()

    if jumlah > 0:
        print(kolom, ":", jumlah, "data mencurigakan")
```

```python
# Memeriksa data duplikat

print("Jumlah baris duplikat:",
      data.duplicated().sum())

print("Jumlah ID transaksi duplikat:",
      data["ID_Transaksi"].duplicated().sum())
```

```python
# Menampilkan seluruh baris yang terduplikasi

duplikat_baris = data[data.duplicated(keep=False)]

display(duplikat_baris)
```

```python
# Menampilkan ID transaksi yang berulang

duplikat_id = data[
    data["ID_Transaksi"].duplicated(keep=False)
]

display(duplikat_id.sort_values("ID_Transaksi"))
```

```python
# Memeriksa inkonsistensi variabel kategorik

kolom_kategorik = [
    'Jenis_Kelamin',
    'Nama_Produk',
    'Kategori_Produk',
    'Metode_Pembayaran',
    'Status_Pesanan',
    'Jenis_Pelanggan'
]

for kolom in kolom_kategorik:
    print("\nVariabel:", kolom)
    print(data[kolom].dropna().unique())
```

```python
# Pemeriksaan statistik deskriptif variabel numerik

kolom_numerik = [
    'Usia_Pelanggan',
    'Jumlah_Pembelian',
    'Harga_Satuan',
    'Total_Penjualan'
]

angka_pemeriksaan = pd.DataFrame()

for kolom in kolom_numerik:
    nilai = data[kolom].astype('string').str.strip()

    nilai = nilai.str.replace(
        r'(?i)^rp\s*', '', regex=True
    )

    nilai = nilai.str.replace('.', '', regex=False)

    angka_pemeriksaan[kolom] = pd.to_numeric(
        nilai, errors='coerce'
    )

display(angka_pemeriksaan.describe().T)
```

```python
# Deteksi outlier menggunakan metode IQR

for kolom in kolom_numerik:

    Q1 = angka_pemeriksaan[kolom].quantile(0.25)
    Q3 = angka_pemeriksaan[kolom].quantile(0.75)

    IQR = Q3 - Q1

    batas_bawah = Q1 - (1.5 * IQR)
    batas_atas = Q3 + (1.5 * IQR)

    outlier = angka_pemeriksaan[
        (angka_pemeriksaan[kolom] < batas_bawah) |
        (angka_pemeriksaan[kolom] > batas_atas)
    ]

    print("\nVariabel:", kolom)
    print("Batas bawah:", batas_bawah)
    print("Batas atas:", batas_atas)
    print("Jumlah kandidat outlier:", len(outlier))

    display(data.loc[outlier.index, ["ID_Transaksi", kolom]])
```

```python
# Memeriksa format dan validitas tanggal

tanggal_periksa = pd.to_datetime(
    data["Tanggal_Transaksi"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Jumlah tanggal tidak valid:",
      tanggal_periksa.isna().sum())

print("\nData tanggal yang tidak valid:")

display(
    data.loc[
        tanggal_periksa.isna(),
        ["ID_Transaksi", "Tanggal_Transaksi"]
    ]
)
```

```python
# Memeriksa kesesuaian total penjualan

cek_total = pd.DataFrame()

cek_total["ID_Transaksi"] = data["ID_Transaksi"]

cek_total["Jumlah_Pembelian"] = angka_pemeriksaan["Jumlah_Pembelian"]

cek_total["Harga_Satuan"] = angka_pemeriksaan["Harga_Satuan"]

cek_total["Total_Penjualan"] = angka_pemeriksaan["Total_Penjualan"]

# Menghitung total yang seharusnya
cek_total["Total_Seharusnya"] = (
    cek_total["Jumlah_Pembelian"] *
    cek_total["Harga_Satuan"]
)

# Memilih data yang angka-angkanya tersedia
valid = cek_total[
    ["Jumlah_Pembelian", "Harga_Satuan", "Total_Penjualan"]
].notna().all(axis=1)

hasil_cek = cek_total[valid].copy()

# Membandingkan total aktual dengan hasil perhitungan
hasil_cek["Sesuai"] = np.isclose(
    hasil_cek["Total_Penjualan"],
    hasil_cek["Total_Seharusnya"]
)

print("Jumlah transaksi tidak sesuai:",
      (~hasil_cek["Sesuai"]).sum())

display(
    hasil_cek[~hasil_cek["Sesuai"]]
)
```

```python

# Memperbaiki pemeriksaan data yang gagal dikonversi

# Menyamakan indeks kedua dataset
data_asli_pos = data.reset_index(drop=True)
data_clean_pos = data_clean.reset_index(drop=True)

for kolom in kolom_numerik:

    raw = (
        data_asli_pos[kolom]
        .astype("string")
        .str.strip()
    )

    # Membersihkan format angka
    raw_bersih = raw.str.replace(
        r"(?i)^rp\s*", "", regex=True
    )

    raw_bersih = raw_bersih.str.replace(
        ".", "", regex=False
    )

    # Memeriksa apakah teks dapat diubah menjadi angka
    hasil_parse = pd.to_numeric(
        raw_bersih, errors="coerce"
    )

    # Menandai data yang gagal dikonversi
    raw_lower = raw.str.lower().fillna("")

    mask_gagal = (
        data_clean_pos[kolom].isna()
        & raw.notna()
        & raw_lower.ne("")
        & ~raw_lower.isin(["-", "unknown", "n/a"])
        & hasil_parse.isna()
    )

    if mask_gagal.any():

        print("\nVariabel:", kolom)

        display(
            data_asli_pos.loc[
                mask_gagal,
                ["ID_Transaksi", kolom]
            ]
        )

print("\nPemeriksaan selesai.")
```

```python

# Memeriksa missing value terbaru

kolom_asli = [
    "ID_Transaksi",
    "Tanggal_Transaksi",
    "Jenis_Kelamin",
    "Usia_Pelanggan",
    "Nama_Produk",
    "Kategori_Produk",
    "Jumlah_Pembelian",
    "Harga_Satuan",
    "Total_Penjualan",
    "Metode_Pembayaran",
    "Status_Pesanan",
    "Jenis_Pelanggan"
]

print("Jumlah missing value setiap variabel:")

display(data_clean[kolom_asli].isnull().sum())

print("\nTotal seluruh missing value:")

print(data_clean[kolom_asli].isnull().sum().sum())
```

```python

# Mengisi missing value transaksi berdasarkan rumus

kolom_transaksi = [
    "Jumlah_Pembelian",
    "Harga_Satuan",
    "Total_Penjualan"
]

# Memastikan tipe data numerik
for kolom in kolom_transaksi:
    data_clean[kolom] = pd.to_numeric(
        data_clean[kolom], errors="coerce"
    )

# 1. Mengisi total penjualan yang kosong
mask_total = (
    data_clean["Total_Penjualan"].isna() &
    data_clean["Jumlah_Pembelian"].notna() &
    data_clean["Harga_Satuan"].notna()
)

idx_total = data_clean.index[mask_total]

data_clean.loc[idx_total, "Total_Penjualan"] = (
    data_clean.loc[idx_total, "Jumlah_Pembelian"] *
    data_clean.loc[idx_total, "Harga_Satuan"]
)

# 2. Mengisi jumlah pembelian yang kosong
mask_jumlah = (
    data_clean["Jumlah_Pembelian"].isna() &
    data_clean["Harga_Satuan"].notna() &
    data_clean["Total_Penjualan"].notna() &
    (data_clean["Harga_Satuan"] > 0)
)

idx_jumlah = data_clean.index[mask_jumlah]

rasio_jumlah = (
    data_clean.loc[idx_jumlah, "Total_Penjualan"].astype(float) /
    data_clean.loc[idx_jumlah, "Harga_Satuan"].astype(float)
)

valid_jumlah = (
    (rasio_jumlah > 0) &
    np.isclose(
        rasio_jumlah.to_numpy(),
        np.round(rasio_jumlah.to_numpy())
    )
)

idx_jumlah_valid = rasio_jumlah.index[valid_jumlah]

data_clean.loc[
    idx_jumlah_valid, "Jumlah_Pembelian"
] = np.round(rasio_jumlah.loc[idx_jumlah_valid]).astype(int)

# 3. Mengisi harga satuan yang kosong
mask_harga = (
    data_clean["Harga_Satuan"].isna() &
    data_clean["Jumlah_Pembelian"].notna() &
    data_clean["Total_Penjualan"].notna() &
    (data_clean["Jumlah_Pembelian"] > 0)
)

idx_harga = data_clean.index[mask_harga]

rasio_harga = (
    data_clean.loc[idx_harga, "Total_Penjualan"].astype(float) /
    data_clean.loc[idx_harga, "Jumlah_Pembelian"].astype(float)
)

valid_harga = (
    (rasio_harga > 0) &
    np.isclose(
        rasio_harga.to_numpy(),
        np.round(rasio_harga.to_numpy())
    )
)

idx_harga_valid = rasio_harga.index[valid_harga]

data_clean.loc[
    idx_harga_valid, "Harga_Satuan"
] = np.round(rasio_harga.loc[idx_harga_valid]).astype(int)

# Menampilkan hasil perbaikan
idx_perbaikan = pd.Index(
    list(idx_total) +
    list(idx_jumlah_valid) +
    list(idx_harga_valid)
).unique()

print("Data transaksi setelah pengisian:")

display(
    data_clean.loc[
        idx_perbaikan,
        [
            "ID_Transaksi",
            "Jumlah_Pembelian",
            "Harga_Satuan",
            "Total_Penjualan"
        ]
    ]
)

print("\nMissing value setelah pengisian:")

display(data_clean[kolom_transaksi].isnull().sum())
```

```python

# Mengisi missing value variabel kategorik menggunakan modus

kolom_kategorik = [
    "Jenis_Kelamin",
    "Nama_Produk",
    "Kategori_Produk",
    "Metode_Pembayaran",
    "Status_Pesanan",
    "Jenis_Pelanggan"
]

print("HASIL PENGISIAN MISSING VALUE KATEGORIK")

for kolom in kolom_kategorik:

    # Menghitung modus
    modus = data_clean[kolom].mode(dropna=True)[0]

    # Menghitung missing sebelum pengisian
    sebelum = data_clean[kolom].isna().sum()

    # Mengisi missing value
    data_clean[kolom] = data_clean[kolom].fillna(modus)

    # Menghitung missing setelah pengisian
    sesudah = data_clean[kolom].isna().sum()

    print("\nVariabel:", kolom)
    print("Modus:", modus)
    print("Missing sebelum:", sebelum)
    print("Missing sesudah:", sesudah)

print("\nPengisian variabel kategorik selesai!")
```

```python

# =====================================
# VALIDASI KESELURUHAN DATASET
# =====================================

kolom_asli = [
    "ID_Transaksi",
    "Tanggal_Transaksi",
    "Jenis_Kelamin",
    "Usia_Pelanggan",
    "Nama_Produk",
    "Kategori_Produk",
    "Jumlah_Pembelian",
    "Harga_Satuan",
    "Total_Penjualan",
    "Metode_Pembayaran",
    "Status_Pesanan",
    "Jenis_Pelanggan"
]

# Membuat dataset akhir sementara
data_final = data_clean[kolom_asli].copy()

# Memastikan tipe data
kolom_numerik = [
    "Usia_Pelanggan",
    "Jumlah_Pembelian",
    "Harga_Satuan",
    "Total_Penjualan"
]

for kolom in kolom_numerik:
    data_final[kolom] = pd.to_numeric(
        data_final[kolom], errors="coerce"
    )

data_final["Tanggal_Transaksi"] = pd.to_datetime(
    data_final["Tanggal_Transaksi"], errors="coerce"
)

# Memeriksa jumlah data
print("=== UKURAN DATASET ===")
print("Jumlah baris dan kolom:", data_final.shape)

# Memeriksa missing value
print("\n=== MISSING VALUE ===")
print("Total missing value:",
      data_final.isnull().sum().sum())

# Memeriksa duplikasi
print("\n=== DATA DUPLIKAT ===")
print("Duplikat seluruh baris:",
      data_final.duplicated().sum())

print("ID transaksi duplikat:",
      data_final["ID_Transaksi"].duplicated().sum())

# Memeriksa tipe data
print("\n=== TIPE DATA ===")
display(data_final.dtypes)

# Memeriksa konsistensi transaksi
total_seharusnya = (
    data_final["Jumlah_Pembelian"] *
    data_final["Harga_Satuan"]
)

sesuai = np.isclose(
    data_final["Total_Penjualan"],
    total_seharusnya
)

print("\n=== KONSISTENSI TRANSAKSI ===")
print("Transaksi sesuai:", sesuai.sum())
print("Transaksi tidak sesuai:", (~sesuai).sum())

# Statistik deskriptif
print("\n=== STATISTIK DESKRIPTIF ===")
display(data_final[kolom_numerik].describe().T)

print("\n=== VALIDASI SELESAI ===")
```

```python

# PEMERIKSAAN OUTLIER AKHIR MENGGUNAKAN IQR

kolom_numerik_final = [
    "Usia_Pelanggan",
    "Jumlah_Pembelian",
    "Harga_Satuan",
    "Total_Penjualan"
]

for kolom in kolom_numerik_final:

    Q1 = data_final[kolom].quantile(0.25)
    Q3 = data_final[kolom].quantile(0.75)

    IQR = Q3 - Q1

    batas_bawah = Q1 - (1.5 * IQR)
    batas_atas = Q3 + (1.5 * IQR)

    mask = (
        (data_final[kolom] < batas_bawah) |
        (data_final[kolom] > batas_atas)
    )

    kandidat = data_final.loc[mask]

    print("\nVariabel:", kolom)
    print("Batas bawah:", batas_bawah)
    print("Batas atas:", batas_atas)
    print("Jumlah kandidat outlier:", len(kandidat))

    display(
        kandidat[
            ["ID_Transaksi", "Nama_Produk", kolom]
        ]
    )
```
